## Reconocimiento inicial de la base de datos

```python
import pandas as pd

raiz_f1 = raiz_repo / "F1"
df = pd.read_csv(raiz_f1 / "data" / "raw" / "Atropellos_2023.csv")
print("Dimensiones:", df.shape)
df.head()
```

```python
# Resumen de registro de datos

celdas_en_blanco = df.astype(str).apply(lambda col: col.str.strip().eq("").sum())

resumen = pd.DataFrame({
    "tipo_dato": df.dtypes,
    "registros_no_nulos": df.count(),
    "valores_nulos": df.isnull().sum(),
    "valores_en_cero": (df.select_dtypes(include="number") == 0).sum().reindex(df.columns, fill_value=0),
    "celdas_en_blanco": celdas_en_blanco,
})
resumen
```

**Nota.** En el dataset se registra la ausencia de información como espacio en blanco o "0" en vez NULL.
